# Phase 1: Dataset Understanding
**Bank Marketing Subscription Prediction**

This notebook inspects the `cestwc/bank-marketing` dataset from Hugging Face to understand its structure, feature types, target distribution, and initial data quality.

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
from datasets import load_dataset
from src.data_loader import load_bank_marketing_data

# Load dataset with decoded categorical labels
df = load_bank_marketing_data(use_cache=True)
print(f"Dataset successfully loaded. Shape: {df.shape}")
```

```python
# Display first 10 rows
df.head(10)
```

```python
# Column names and data types
print("Columns:", df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
```

```python
# Identify numerical and categorical features
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = [c for c in df.select_dtypes(include=['object']).columns if c != 'y']

print(f"Numerical Features ({len(num_cols)}): {num_cols}")
print(f"Categorical Features ({len(cat_cols)}): {cat_cols}")
print(f"Target Column: 'y'")
```

```python
# Summary statistics for numerical features
df[num_cols].describe()
```

```python
# Target distribution (Counts & Normalized %)
target_counts = df['y'].value_counts()
target_pct = df['y'].value_counts(normalize=True) * 100
pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_pct.round(2)})
```

```python
# Data quality: Missing values & Duplicates
print("Missing Values (NaNs):")
print(df.isnull().sum())

print(f"\nDuplicate Rows: {df.duplicated().sum()}")
```

```python
# Inspect unique categorical values and 'unknown' category prevalence
for col in cat_cols:
    unknown_count = (df[col] == 'unknown').sum()
    unknown_pct = (unknown_count / len(df)) * 100
    print(f"{col.upper()} (Unique: {df[col].nunique()} | Unknowns: {unknown_count} [{unknown_pct:.2f}%]):")
    print(df[col].value_counts())
    print("-" * 50)
```
